$\Large \text{TO SUMMARIZE}$

The table summarises the testing accuracy across all the 20 models.

```
        | Vanilla |Bahdanau | Dot  | General |Concat
----------------------------------------------------
RNN     |  0.51   |  0.73   | 0.50 |  0.56   | 0.80
----------------------------------------------------
BiRNN   |  0.50   |  0.80   | 0.86 |  0.67   | 0.80
----------------------------------------------------         
LSTM    |  0.71   |  0.86   | 0.64 |  0.85   | 0.83
----------------------------------------------------
BiLSTM  |  0.83   |  0.85   | 0.86 |  0.83   | 0.87
----------------------------------------------------
```



Overall, Bahdanau Attention along with Luong's adjustment to it, was the best performing attention architecture.  
Bidirectional models performed consistently better than or as well as their unidirectional variants.  
Dot attention should be incorporated only into bidirectional variants of RNN and LSTM.
